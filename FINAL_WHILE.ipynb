```python
# First 10 Even number

count = 1
i = 1
while count < 11:
    if i%2 == 0:
        print(i)
        count += 1
    i +=1
```

```python
# Q2.First 10 Odd numbers

count = 1
i = 1
while count < 11:
    if i%2 != 0:
        print(i)
        count += 1
    i +=1
```

```python
# Q10. Write a program to print sum of first 10 Even numbers.
add_even = 0
count = 0
i = 2
while count < 10:
    if i % 2 == 0:
        count += 1
        add_even += i
    i += 1
print(add_even)
```

```python
# Q11. Write a program to print table of a number entered from the user.

n = int(input('enter no: '))
i = 1

while i <= 10:
    print(n*i,end=' ')
    i += 1
```

```python
# Q12. Write a program to print all even numbers that falls between two numbers (exclusive both numbers) entered from the user using while loop.

a = int(input('enter one no: '))
b = int(input('enter another no: '))
a += 1
while a < b:
    if a%2 == 0:
        print(a,end=' ')
    a+=1
```
