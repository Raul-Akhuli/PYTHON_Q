```python
# Q13. Write a program to check whether a number is prime or not using while loop.

n = int(input('Enter a number: '))
if n <=1:
    print('not prime.')
if n == 2:
    print('prime')

flag = True
i = 2
while i<n:
    if n%i == 0:
        flag = False
        break
    i += 1

if flag == True:
    print('Entered number is prime.')
else:
    print('Not Prime')
```

```python
# Q14. Write a program to find the sum of the digits of a number accepted from the user.

n = int(input('Enter no: '))
temp = n
add = 0
while n>0:
    digit = n%10
    add +=digit
    n //= 10
print(f'sum of digits of number {temp} is {add}')
```

```python
# Q15. Write a program to find the product of the digits of a number accepted from the user.

n = int(input('Enter no: '))
temp = n
pro = 1
while n>0:
    digit = n%10
    pro *=digit
    n //= 10
print(f'product of digits of number {temp} is {pro}')
```

```python
# Q16. Write a program to reverse the number accepted from user using while loop.

n = int(input('Enter no: '))
temp = n

rev = 0
while n>0:
    digit = n%10
    rev = rev*10+ digit
    n //= 10
print(f'reverse of digits of number {temp} is {rev}')
```

```python
# Q17. Write a program to display the number names of the digits of a number entered by user, for example if the number is 231 then output should be Two Three One

book = {1:'One',2:'Two',3:'Three',4:'Four',5:'Five',6:'Six',7:'Seven',8:'Eight',9:'Nine',0:'Zero'}

n = int(input('Enter no: '))
out = ''
temp = n

while n>0:
    digit = n%10
    out = book[digit] +' '+ out
    n//=10

print(f'name of digit {temp} is {out}')
```
