```python
# Method concept

class Car:
    model = 'bmw'
    color = 'blue'
    def __init__(self,val1,val2,val3):
        self.model = val1
        self.color = val2
        self.price = val3

car1 = Car('maruti 500','white', 500000)

print(Car.model)
print(Car.color)
print(car1.model)
print(car1.color)
print(car1.price)
```

```python
# Types of Method

class School:
    sc_name = 'Hindu School'
    sc_loc = 'Kolkata'

    def __init__(self,name,roll):
        self.Name = name
        self.Roll = roll
    def display(self):
        print(self.Name, self.Roll)              # object method
    def modify_roll(self,roll):
        self.Roll = roll

    @classmethod
    def disp(cls):
        print(cls.sc_name, cls.sc_loc)
    @classmethod                                 # class method
    def modify_loc(cls, location):
        cls.sc_loc = location

    @staticmethod
    def greeting():
        print('Hello Guys!!!!!')                 # static method

    

ob = School('Raul',46)
# print(ob)  # it only give you the address of the object 


# ob.display()
# ob.modify_roll(33)                             # object method
# ob.display()


# School.disp()
# School.modify_loc('collage street')              # class method
# School.disp()



ob.greeting()                                     # static method
School.greeting()
```

```python
# Inheritance

class A:
    a = 10
    b = 50
class B(A):
    a = 20
    c = 40
    d = 80

ob = A()
ob1 = B()

print(ob.a,ob.b)
print(ob1.a,ob1.c,ob1.d)
```

```python
# constractor and method chaining

class Resume_12:
    def __init__(self,name,roll,per_10,per_12): #----->4
        self.Name = name
        self.Roll = roll
        self.Per_10 = per_10
        self.Per_12 = per_12

    def display(self):#----->d
        print(self.Name, self.Roll, self.Per_10,self.Per_12)#----->e

class Resume_btech(Resume_12):
    def __init__(self,a,b,c,d,e,f):#----->2
        super().__init__(a,b,c,d) # constractor chaining. ----->3
        self.gread_per = e#----->5
        self.cgpa = f#----->6

    def disp(self):#----->b
        super().display() # method chaining. ----->c
        print(self.Name, self.Roll, self.Per_10,self.Per_12,self.gread_per,self.cgpa)#----->f


# ob = Resume_12()                                    # do not have to create the parent class object 
ob1 = Resume_btech('Arnab', 3020, 91,85, 90,9.1) #----->1

ob1.disp()#----->a
```
