```python
n = int(input('enter row: '))
m = int(input('enter column: '))

mid = n//2 +1

for i in range (1, n+1):
    for j in range(1, m+1):
        if i+j >= mid+1 and j-i <= mid-1 and i-j <= mid-1 and i+j <= n+mid:
            print('*',end='')
        else:
            print(' ', end='')
    print()

# i+j >= mid+1 and j-i <= mid-1 and i-j <= mid-1 and i+j <= n+mid
```

```python
# U
n = int(input('enter no of rows: '))

for i in range(1,n+1):
    for j in range(1,n+1):
        if i == n or j == 1 or j == n:
            print('*', end='')
        else:
            print(' ', end='')
    print()
```

```python
# M
n = int(input('enter no of rows: '))
mid = n//2 +1

for i in range(1,n+1):
    for j in range(1,n+1):
        if j == 1 or j == n or (i == j and i <= mid) or (i+j == n+1 and i <= mid):
            print('*', end='')
        else:
            print(' ', end='')
    print()
```

```python
# G
n = int(input('enter no of rows: '))
mid = n//2 +1

for i in range(1,n+1):
    for j in range(1,n+1):
        if i == 1 or j == 1 or i == n or (j == n and j >= mid) or (i >= mid and i == mid):
            print('*', end='')
        else:
            print(' ', end='')
    print()
```
