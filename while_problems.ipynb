```python
# Q17. Write a program to display the number names of the digits of a number entered by user, for example if the number is 231 then output should be Two Three One

d={
    1:'one',2:'two',3:'three'
}

num = int(input('enter the number: '))
out = ''

while num > 0:
    digit = num%10
    out = d[digit] + ' ' + out
    num //= 10

print(out)
```

```python
# Q21. 
# Write a program to add first n terms of the following series using a while loop:
# 1/1! + 1/2! + 1/3! + …….. + 1/n!

num = int(input('enter no: '))
i = 1
add = 0
fact = 1

while i <= num:
    fact = i*fact
    add += 1/fact
    i += 1

print(add)
```

```python
# Q23. Count positive and negative numbers until 0 is entered

pos = 0
neg = 0

while True:
    num = int(input("Enter number (0 to stop): "))
    if num == 0:
        break
    elif num > 0:
        pos += 1
    else:
        neg += 1

print("Positive =", pos)
print("Negative =", neg)
```

```python
# Q25. Decimal to Binary

n = int(input("Enter decimal number: "))
binary = ""

while n > 0:
    binary = str(n % 2) + binary
    n //= 2

print("Binary =", binary)
```

```python
# Q26. Binary to Decimal

binary = input("Enter binary number: ")
i = len(binary) - 1
power = 0
decimal = 0

while i >= 0:
    decimal += int(binary[i]) * (2 ** power)
    power += 1
    i -= 1

print("Decimal =", decimal)
```

```python
# Q27. Palindrome number

n = int(input("Enter number: "))
temp = n
rev = 0

while temp > 0:
    rev = rev * 10 + temp % 10
    temp //= 10

if n == rev:
    print("Palindrome")
else:
    print("Not Palindrome")
```

```python
# Q30. Largest and smallest among 10 numbers

i = 1
num = int(input("Enter number: "))
largest = smallest = num

while i < 10:
    num = int(input("Enter number: "))
    if num > largest:
        largest = num
    if num < smallest:
        smallest = num
    i += 1

print("Largest =", largest)
print("Smallest =", smallest)
```

```python
# Q31. Sum of odd and even numbers between two numbers

a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

even_sum = 0
odd_sum = 0

while a <= b:
    if a % 2 == 0:
        even_sum += a
    else:
        odd_sum += a
    a += 1

print("Even Sum =", even_sum)
print("Odd Sum =", odd_sum)
```

```python
# Q39. Sum: 1 - 4 + 9 - 16 + 25 ...

n = int(input("Enter n: "))
i = 2
s = 1

while i <= n:
    if i % 2 == 0:
        s -= i * i
    else:
        s += i * i
    i += 1

print("Sum =", s)
```

```python
# Q60.Write a program to get the following output
# input='hai hello good morning'
# output={‘hai’:’a’ , ‘hello’: ‘l’ , ‘good’:’gd’ , ‘morning’:’n’}

st='hai hello good morning'
word = st.split()
i = 0
d={}
while i < len(word):
    w = word[i]
    if len(w) % 2 == 0:
        d[w] = w[0] + w[-1]
    else:
        middle = len(w)//2
        d[w] = w[middle]
    i += 1
print(d)
```
